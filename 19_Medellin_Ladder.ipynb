# The same ladder on an independent public dataset

The ladder of notebooks 13 to 16 (what the predictor knows against the margin it needs) comes from one office and six links. Here the same procedure runs on the public urban measurements of González-Palacio et al., the data behind the machine-learning path-loss models for LoRaWAN of Medellín: four fixed end nodes, 930,753 uplinks from October 2021 to March 2022, roughly one a minute, with temperature, humidity, pressure, PM2.5, the spreading factor and the SNR.

Source: the companion data descriptor, https://doi.org/10.3390/data8010004, and the repository `magonzalezudem/MDPI_LoRaWAN_Dataset_With_Environmental_Variables`. The file `LoRaWAN_PathLossMeasurements.csv` from that repository goes into `Data_Files`. It is read unchanged, without the Mahalanobis filter of the original paper, which uses the path loss itself among its variables.

* Same chronological split as notebook 00: the first 80 % of the packets in time for training, five forward-chaining validation blocks inside it, the last 20 % as hold-out. The timestamps are taken as local time, UTC-5.
* The history features are those of `history_features.build`, with the clock in the local time zone. The environmental inputs are the four sensors of the dataset.
* **Static inputs**: distance, frequency, temperature, humidity, pressure, PM2.5. **Static inputs with the SNR of the same packet**: an input of the models published on these data; it is measured on the packet being predicted, so it is known only after reception. History rungs as in notebook 13: H1 link history; H4 adds the same-channel history, the clock and the spreading factor; H6 adds the SNR of earlier packets and the sensors; H7 adds the link level and the distance.
* The history rungs share one LightGBM setting, tuned on rung H4 as in notebook 13; the static input sets, with and without the SNR of the predicted packet, are tuned on their own by the same search, so that no input set is scored with settings chosen for another. The 99 % margin is one shift from the out-of-fold scores of the validation blocks. Notebook 16's four main methods are repeated on rung H4.
* A link the model has never seen, as in notebook 13: fitted on three links, margin calibrated on their later packets, tested on the fourth link's hold-out, with the margin kept fixed or adapted to the link's own misses.

```python
import json
import time
import numpy as np
import pandas as pd
import lightgbm as lgb
from sklearn.model_selection import TimeSeriesSplit
from skopt import BayesSearchCV
from skopt.space import Integer, Real
from history_features import build, H1, H2, H3, H4, H5
from margin_tools import pinball, shift_pooled, shift_groups, aci, spread_groups, day_codes, boot_stat, transitions, clustering_ratio

RANDOM_STATE = 42
TAU = 0.99
```

```python
# The public file, as released
raw = pd.read_csv("Data_Files/LoRaWAN_PathLossMeasurements.csv")
df = pd.DataFrame({"device_id": raw["device_id"],
                   "t": pd.to_datetime(raw["timestamp"], dayfirst=True, format="mixed").dt.tz_localize("America/Bogota").dt.tz_convert("UTC"),
                   "PL": raw["experimental_pl"], "frequency": raw["frequency"] / 1e6, "SF": raw["sf"], "snr": raw["snr"], "distance": raw["distance"],
                   "temperature": raw["temperature"], "rh": raw["rh"], "bp": raw["bp"], "pm2_5": raw["pm2_5"]})
df = df.sort_values("t").reset_index(drop=True)

# The chronological split and the forward-chaining folds of notebook 00 (fold -1 = first training block, 0-4 = validation blocks, 9 = hold-out)
split = int(len(df) * 0.8)
fold = np.full(len(df), 9)
fold[:split] = -1
for k, (_, va) in enumerate(TimeSeriesSplit(n_splits=5).split(df.iloc[:split])):
    fold[va] = k
df["fold"] = fold
print(df.groupby("device_id").agg(packets=("PL", "size"), first=("t", "min"), last=("t", "max"), distance=("distance", "first"), PL_mean=("PL", "mean"), PL_sd=("PL", "std")).round(2))
print(f"training {split:,} packets up to {df['t'].iloc[split - 1]}, hold-out {len(df) - split:,} packets from {df['t'].iloc[split]}; hold-out links: {sorted(df.loc[fold == 9, 'device_id'].unique())}")
```

```python
# History features and rungs
d = build(df, tz="America/Bogota").sort_values("t").reset_index(drop=True)
fold = d["fold"].to_numpy()
SENSORS = ["temperature", "rh", "bp", "pm2_5"]
rungs = {"H1": H1, "H4": H4, "H6": H5 + SENSORS, "H7": H5 + SENSORS + ["a", "distance"]}
static = ["distance", "frequency"] + SENSORS
ok = (d["y"].notna() & d[rungs["H7"]].notna().all(axis=1)).to_numpy()
y, a, PL = d["y"].to_numpy(), d["a"].to_numpy(), d["PL"].to_numpy()
link, sf, work, s20 = d["device_id"].to_numpy(), d["SF"].to_numpy().astype(int), d["work"].to_numpy().astype(int), d["s20"].to_numpy()
day = day_codes(d["t"])
val, ev = ok & (fold >= 0) & (fold < 9), ok & (fold == 9)
X_all = d[sorted(set(sum(rungs.values(), []) + static + ["snr"]))]
X_all = X_all.fillna(X_all[ok & (fold < 9)].median())
print(f"packets with an anchor and a complete input: {ok.sum():,} of {len(d):,}; validation {val.sum():,}; hold-out {ev.sum():,}")
```

```python
# One light search per input set on every tenth training row, folds kept, as in notebook 13: rung H4 gives the settings of every history rung,
# and each static input set is searched on its own in the same space, so that it is not scored with settings chosen for the history inputs
sub = np.flatnonzero(ok & (fold < 9))[::10]
cv_sub = [(np.flatnonzero(fold[sub] < k), np.flatnonzero(fold[sub] == k)) for k in range(5)]
space = {"num_leaves": Integer(15, 255), "learning_rate": Real(0.01, 0.2, prior="log-uniform"), "n_estimators": Integer(200, 1500), "min_child_samples": Integer(5, 1000),
         "subsample": Real(0.6, 1.0), "colsample_bytree": Real(0.5, 1.0), "reg_lambda": Real(0.1, 30, prior="log-uniform")}

def search(cols, target):
    t0 = time.time()
    bs = BayesSearchCV(lgb.LGBMRegressor(subsample_freq=1, random_state=RANDOM_STATE, n_jobs=4, verbose=-1), space, n_iter=20, cv=cv_sub, scoring="neg_root_mean_squared_error",
                       n_jobs=8, n_points=2, random_state=RANDOM_STATE, refit=False)
    bs.fit(X_all[cols].to_numpy()[sub], target[sub])
    best = dict(bs.best_params_)
    print(f"search RMSE {-bs.best_score_:.3f} dB after {time.time() - t0:.0f} s: {best}")
    return best

best = search(rungs["H4"], y)
json.dump(best, open("CV_Results/medellin_best_params.json", "w"), indent=1, default=float)
params = dict(subsample_freq=1, random_state=RANDOM_STATE, n_jobs=8, verbose=-1, **best)
static_sets = {"static inputs": static, "static inputs + SNR of the same packet": static + ["snr"]}
static_best = {}
for name, cols in static_sets.items():
    print(name, end=": ")
    static_best[name] = search(cols, PL)
json.dump(static_best, open("CV_Results/medellin_static_best_params.json", "w"), indent=1, default=float)
static_params = {name: dict(subsample_freq=1, random_state=RANDOM_STATE, n_jobs=8, verbose=-1, **p) for name, p in static_best.items()}

def oof_and_test(make, cols, target):
    X = X_all[cols].to_numpy()
    p = np.full(len(d), np.nan)
    for k in range(5):
        tr, va = ok & (fold < k), ok & (fold == k)
        p[va] = make().fit(X[tr], target[tr]).predict(X[va])
    p[ev] = make().fit(X[ok & (fold < 9)], target[ok & (fold < 9)]).predict(X[ev])
    return p
```

```python
# The ladder: predicted path loss for every rung, and the references that need no model
pred = {}
t0 = time.time()
mean = np.full(len(d), np.nan)                                       # the mean of each link over the earlier training packets
for k in list(range(5)) + [9]:
    earlier = d[ok & (fold < k)] if k < 9 else d[ok & (fold < 9)]
    mean[fold == k] = d.loc[fold == k, "device_id"].map(earlier.groupby("device_id")["PL"].mean())
pred["per-link mean"] = mean
pred["previous hour mean"] = a
for name, cols in static_sets.items():
    pred[name] = oof_and_test(lambda: lgb.LGBMRegressor(**static_params[name]), cols, PL)
for name, cols in rungs.items():
    pred[f"rung {name}"] = a + oof_and_test(lambda: lgb.LGBMRegressor(**params), cols, y)
print(f"{time.time() - t0:.0f} s")

te = np.flatnonzero(ev)
dd = pd.factorize(day[te])[0]
rmse = lambda e: float(np.sqrt(np.mean(np.square(e))))
rows = []
for name, p in pred.items():
    e = PL - p
    c = shift_pooled(e[val & np.isfinite(e)], TAU)
    r = boot_stat(dd, np.column_stack([e[te] ** 2, np.ones(len(te))]), lambda T: np.sqrt(T[:, 0] / T[:, 1]))
    rows.append({"inputs": name, "hold-out RMSE": r[0], "lo": r[1], "hi": r[2], "validation RMSE": rmse(e[val & np.isfinite(e)]), "99 % shift dB": c, "coverage": float((e[te] <= c).mean()), "pinball x 1000": 1000 * pinball(e[te], c, TAU).mean()})
ladder = pd.DataFrame(rows).set_index("inputs")
ladder.to_csv("CV_Results/medellin_ladder.csv")
display(ladder.round(3))
```

```python
# The SNR of the predicted packet, link by link: the SNR at which each link operates and the slope of its path loss on that SNR (training packets),
# against the hold-out RMSE of the static model without and with that SNR, with 95 % intervals from resampling days
without, with_ = "static inputs", "static inputs + SNR of the same packet"
snr = d["snr"].to_numpy()
rows = []
for l in ["all"] + sorted(np.unique(link[te])):
    i = te if l == "all" else te[link[te] == l]
    j = ok & (fold < 9) & np.isfinite(snr)
    if l != "all":
        j &= link == l
    e0, e1 = (PL - pred[without])[i], (PL - pred[with_])[i]
    r = boot_stat(pd.factorize(day[i])[0], np.column_stack([e1 ** 2, e0 ** 2, np.ones(len(i))]), lambda T: np.sqrt(T[:, 0] / T[:, 2]) - np.sqrt(T[:, 1] / T[:, 2]))
    rows.append({"link": l, "hold-out packets": len(i), "median SNR dB": np.median(snr[j]), "SNR 10 %": np.percentile(snr[j], 10), "SNR 90 %": np.percentile(snr[j], 90),
                 "path loss per dB of SNR": np.polyfit(snr[j], PL[j], 1)[0] if l != "all" else np.nan,
                 "RMSE without": rmse(e0), "RMSE with": rmse(e1), "difference": r[0], "lo": r[1], "hi": r[2]})
leak = pd.DataFrame(rows).set_index("link")
leak.to_csv("CV_Results/medellin_snr_leak.csv")
display(leak.round(3))
```

```python
# Rung H4 with the margin built four ways, as in notebook 16: a regression with one margin, a regression with a margin per state, a quantile model with one shift, a quantile model with an adaptive shift
X4 = rungs["H4"]
reg = pred["rung H4"] - a
q = oof_and_test(lambda: lgb.LGBMRegressor(objective="quantile", alpha=TAU, **params), X4, y)
windows = [1, 2, 3, 4, 9]
margin, cuts, rows_of = {}, {}, {}
for w in windows:
    t_ = np.flatnonzero(ok & (fold == w))
    cal = np.flatnonzero(ok & (fold >= 0) & (fold < w))
    cut = np.quantile(s20[cal], [1 / 3, 2 / 3])
    rows_of[w], cuts[w] = t_, cut
    g1 = lambda i: (sf[i] - 7) * 2 + work[i]
    g2 = lambda i: g1(i) * 3 + spread_groups(s20[i], cut)
    for kind, b in (("reg", reg), ("q", q)):
        score = y[cal] - b[cal]
        margin[(w, f"{kind} pooled")] = b[t_] + shift_pooled(score, TAU)
        if kind == "reg":
            margin[(w, "reg state table")] = b[t_] + shift_groups(score, g2(cal), g2(t_), TAU)
        else:
            m = np.empty(len(t_))
            for l in np.unique(link[t_]):
                i = np.flatnonzero(link[t_] == l)
                m[i] = aci(b[t_][i], y[t_][i], score, TAU, (1 - TAU) / 20)
            margin[(w, "q ACI")] = m
methods = ["reg pooled", "reg state table", "q pooled", "q ACI"]

rows = []
for w in windows:
    t_ = rows_of[w]
    for method in methods:
        m = margin[(w, method)]
        miss = y[t_] > m
        states = {"all": np.ones(len(t_), bool), "other hours": work[t_] == 0, "working hours": work[t_] == 1, "calm spread": spread_groups(s20[t_], cuts[w]) == 0,
                  "volatile spread": spread_groups(s20[t_], cuts[w]) == 2}
        states.update({f"SF{k}": sf[t_] == k for k in np.unique(sf[t_]) if (sf[t_] == k).sum() >= 1000})
        states.update({str(k): link[t_] == k for k in np.unique(link[t_]) if (link[t_] == k).sum() >= 1000})
        for g, sel in states.items():
            rows.append(dict(window=w, method=method, state=g, n=int(sel.sum()), coverage=1 - miss[sel].mean(), margin=m[sel].mean(), pinball=1000 * pinball(y[t_][sel], m[sel], TAU).mean()))
margins = pd.DataFrame(rows)
margins.to_csv("CV_Results/medellin_margins.csv", index=False)
print("coverage at the 99 % target by window and method, all packets")
display(margins[margins.state == "all"].pivot(index="method", columns="window", values="coverage").loc[methods].round(4))
print("hold-out coverage by state")
display(margins[margins.window == 9].pivot(index="state", columns="method", values="coverage")[methods].round(4))
print("hold-out mean margin above the anchor (dB) and pinball loss x 1000, all packets")
display(margins[(margins.window == 9) & (margins.state == "all")].set_index("method")[["margin", "pinball"]].loc[methods].round(3))
```

```python
# Independence of the misses on the hold-out, and the difference between the margins with 95 % intervals from resampling days
t_ = rows_of[9]
dd = pd.factorize(day[t_])[0]
rows = []
for method in methods:
    T = transitions(d["t"].iloc[t_], link[t_], y[t_] > margin[(9, method)])
    r = boot_stat(dd, T, clustering_ratio)
    rows.append(dict(method=method, miss_clustering=r[0], lo=r[1], hi=r[2]))
independence = pd.DataFrame(rows).set_index("method")
display(independence.round(2))
rows = []
for m1, m2 in [("q pooled", "reg pooled"), ("q pooled", "reg state table"), ("q ACI", "q pooled")]:
    diff = pinball(y[t_], margin[(9, m1)], TAU) - pinball(y[t_], margin[(9, m2)], TAU)
    r = boot_stat(dd, np.column_stack([diff, np.ones(len(t_))]), lambda T: 1000 * T[:, 0] / T[:, 1])
    rows.append(dict(a=m1, b=m2, pinball_difference=r[0], lo=r[1], hi=r[2]))
display(pd.DataFrame(rows).round(3))
```

```python
# A link the model has never seen, as in notebook 13: fit on the other three links with validation folds 0 to 3, calibrate the 99 % margin on their fold 4 and test on the
# fourth link's hold-out, with the margin either kept as calibrated or adapted to the link's own misses as it goes (adaptive conformal inference, as in cell 6);
# the reference with the link in training is the same model fitted on folds 0 to 3 of all four links, so that the two differ only by the link
rows = []
for name, cols, target in (("static inputs", static, PL), ("rung H1", rungs["H1"], y), ("rung H4", rungs["H4"], y)):
    X = X_all[cols].to_numpy()
    make = lambda: lgb.LGBMRegressor(**static_params.get(name, params))
    fit_all = ok & (fold >= 0) & (fold < 4)
    m_all = make().fit(X[fit_all], target[fit_all])
    for l in sorted(np.unique(link)):
        other, this = link != l, link == l
        fit_m, cal_m, test_m = ok & other & (fold >= 0) & (fold < 4), ok & other & (fold == 4), ev & this
        m = make().fit(X[fit_m], target[fit_m])
        b, e_cal = m.predict(X[test_m]), target[cal_m] - m.predict(X[cal_m])
        e_test = target[test_m] - b
        c = shift_pooled(e_cal, TAU)
        adaptive = aci(b, target[test_m], e_cal, TAU, (1 - TAU) / 20)
        rows.append({"inputs": name, "link": l, "hold-out packets": int(test_m.sum()), "RMSE unseen": rmse(e_test), "bias unseen": float(e_test.mean()),
                     "RMSE with the link in training": rmse(target[test_m] - m_all.predict(X[test_m])), "99 % shift from the other links": c, "coverage": float((e_test <= c).mean()),
                     "adaptive margin": float((adaptive - b).mean()), "adaptive coverage": float((target[test_m] <= adaptive).mean())})
unseen = pd.DataFrame(rows).set_index(["inputs", "link"])
unseen.to_csv("CV_Results/medellin_unseen_link.csv")
display(unseen.round(3))
```

Reading of the results.

* The ladder on the public urban data: the hold-out RMSE falls from 2.12 dB with the static inputs to 1.54 with the link's history, 1.33 with the same-channel history, the clock and the spreading factor, and 1.31 and 1.30 with the earlier SNR, the sensors and the geometry. The 99 % margin falls from 5.5 to 3.7, 3.0, 2.9 and 2.9 dB, which is 50 to 52 % of the per-link mean's margin (5.7 dB) from rung H4 on. In the office the LightGBM reaches 41 % of the per-link mean's margin at rung H4 and after it (the six families span 41 to 52 %), under the same single-shift rule. In both datasets the later rungs add little.
* The static inputs count for more here than in the office. Here they beat the per-link mean by 0.34 dB, 14 %, with the interval of the static RMSE from 2.03 to 2.22 dB. In the office the static LightGBM is 0.18 dB, 4 %, better than its per-link mean.
* The SNR of the same packet, added to the static inputs with the static model tuned anew for that input set, lowers the RMSE from 2.12 to 1.48 dB (-0.64 dB, interval -0.73 to -0.56), close to the 1.54 dB of the link's own history and below the 1.57 dB published for the environment-aware models on these data. The gain sits on the two links that operate below 0 dB median SNR, EN3 (-12.8 dB) and EN4 (-1.5 dB), whose path loss changes by 1.0 and 1.1 dB per dB of SNR: -0.89 and -1.15 dB. On EN1, at 7 dB median SNR and 0.2 dB per dB, it is -0.05 dB; EN2, with 673 hold-out packets, -0.53 dB. The SNR of a packet is not known before the packet is received, so this input cannot serve a link budget.
* The refinements of the calibration matter little here. The data hold one regime, spreading factors 7 to 10 throughout. The pooled shift covers between 98.8 and 99.3 % in the five windows, and adaptive conformal inference holds 99.0 to 99.06 %. The quantile model lowers the pinball loss by 0.82 per thousand against the regression with one margin (interval 0.54 to 1.12, 2.2 %) and by 0.40 against the margin per state (interval 0.17 to 0.63, 1.1 %). Misses cluster less with the quantile model and the adaptive shift, 1.7 (interval 0.9 to 2.6), than with the regression and one margin, 2.9 (interval 2.0 to 3.8).
* A link the model has never seen. Fitted on blocks 0 to 3 of three links, the static model misses the fourth by 4.5 to 8.2 dB RMSE with biases of -6.7 to +7.9 dB, against 1.6 to 2.2 dB for the same model fitted on the same blocks of all four links: the distance does not determine the level of a link the model has not seen. With the link's history the same fit predicts the fourth link within 0.02 to 0.06 dB of the fit that included it at rung H1 (1.33 to 1.77 dB) and within 0.08 to 0.30 dB at rung H4 (1.22 to 1.67 dB); in the office, against the same kind of reference, -0.01 to +0.04 dB at rung H4. The 99 % margin calibrated on the other three links covers 96.3 to 98.9 % of the unseen link's hold-out packets at rung H4, except EN3: 89.5 %. EN3 transmits at SF10 only while the other three cycle through SF7 to SF10, so the model carries an SF10 offset fitted on the other links that EN3's own anchor already contains; the bias is +0.7 dB. The adaptive shift, fed the link's own misses, brings every history model to 98.2 to 99.0 % (EN3 99.0 % with a 4.4 dB margin). EN2 has 673 hold-out packets, eleven hours, because its campaign ended on 16 February 2022, local time.
* Limits. Four links, one outdoor urban setting, a hold-out of six weeks, and timestamps taken as local time. The office margins of notebook 15 and the margins here are the same single shift from the out-of-fold scores, so the two ladders are compared under one rule.
